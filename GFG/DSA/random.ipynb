```python
#: Randomly Select from List, String, and Tuple
import random

a = [1, 2, 3, 4, 5, 6]
print(random.choice(a))

s = "geeks"
print(random.choice(s))

tup = (1, 2, 3, 4, 5)
print(random.choice(tup))
```

```python
 # Select Multiple Unique Random Items
from random import sample

a = [1, 2, 3, 4, 5, 6]
print(sample(a,3))

s = "geeks"
print(sample(s,2))

tup = (1, 2, 3, 4, 5)
print(sample(tup,3))
```

```python
# : Shuffle Elements in a List

import random
a = [1, 2, 3, 4, 5]

random.shuffle(a)
print("After shuffle : ")
print(a)

random.shuffle(a)
print("\nSecond shuffle : ")
print(a)
```
