```python
import mysql.connector

database = mysql.connector.connect(
    host="localhost",
    user="root",
    password="password",
    database="gfg"
)

cursorObject = database.cursor()


insert = "INSERT INTO student (id, name, branch, section, age) VALUES (%s, %s, %s, %s, %s)"

val = [
    (10, "Nikhil", "CSE", "A", 18),
    (11, "Nisha", "CSE", "A", 18),
    (17, "Rohan", "MAE", "B", 20),
    (13, "Amit", "ECE", "A", 21),
    (14, "Anil", "MAE", "B", 20),
    (15, "Megha", "ECE", "A", 22),
    (16, "Sita", "CSE", "A", 19)
]


cursorObject.executemany(insert, val)
database.commit()

print(cursorObject.rowcount, "records inserted.")
```
