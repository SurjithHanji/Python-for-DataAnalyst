```python
class Person(object):
  def __init__(self, name, id):
    self.name = name
    self.id = id

  def Display(self):
    print(self.name, self.id)

# Driver code
emp = Person("Elon", 102)
emp.Display()
```

```python
class Emp(Person):
  def Print(self):
    print("Emp class called")
    
Emp_details = Emp("Mayank", 103)
Emp_details.Display() # calling parent class function
Emp_details.Print() # Calling child class function
```

```python
# Parent Class: Person
class Person:
    def __init__(self, name, idnumber):
        self.name = name
        self.idnumber = idnumber

    def display(self):
        print(self.name)
        print(self.idnumber)

# Child Class: Employee
class Employee(Person):
    def __init__(self, name, idnumber):
        super().__init__(name, idnumber)
```

```python
# Parent Class: Person
class Person:
    def __init__(self, name, idnumber):
        self.name = name
        self.idnumber = idnumber

    def display(self):
        print(self.name)
        print(self.idnumber)

# Child Class: Employee
class Employee(Person):
    def __init__(self, name, idnumber, salary, post):
        super().__init__(name, idnumber)
        self.salary = salary
        self.post = post
```
