```python
sns.jointplot(x='engine-size',y='horsepower',data=auto)
plt.show()
```

```python
auto['engine-size'] = pd.to_numeric(auto['engine-size'], errors='coerce')
auto['horsepower'] = pd.to_numeric(auto['horsepower'], errors='coerce')

# Drop NaN values (optional, for clean plotting)
auto = auto.dropna(subset=['engine-size', 'horsepower'])

# Now plot
sns.jointplot(x='engine-size', y='horsepower', data=auto, kind="hex")
plt.show()
```

```python
sns.jointplot(x='engine-size', y='horsepower', data=auto, kind="reg")
plt.show()
```

```python
sns.pairplot(auto[['wheel-base','engine-size','horsepower']])
```
